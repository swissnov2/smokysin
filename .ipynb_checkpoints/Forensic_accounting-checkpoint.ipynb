```python
import pandas as pd
import numpy as np

# Creating a mock financial ledger with hidden fraud patterns
data = {
    'transaction_id': [f'TXN_{i:03d}' for i in range(1, 11)],
    'account_id': ['ACC_101', 'ACC_102', 'ACC_101', 'ACC_103', 'ACC_101', 'ACC_104', 'ACC_105', 'ACC_106', 'ACC_101', 'ACC_107'],
    'vendor_name': ['Alpha Corp', 'Beta Ltd', 'Alpha Corp', 'Gamma Inc', 'Delta Services', 'Omega Co', 'Zeta LLC', 'Sigma Corp', 'Alpha Corp', 'Theta Press'],
    'destination_bank_acc': ['BANK_9999', 'BANK_1111', 'BANK_9999', 'BANK_2222', 'BANK_9999', 'BANK_3333', 'BANK_4444', 'BANK_5555', 'BANK_9999', 'BANK_6666'],
    'amount_inr': [49500, 120000, 49800, 15000, 49200, 230000, 8500, 310000, 49600, 45000]
}

df = pd.DataFrame(data)
print("--- ORIGINAL FINANCIAL AUDIT LEDGER ---")
print(df)

print("\n=== RUNNING FORENSIC INVESTIGATION FILTERS ===\n")

# DETECTOR 1: Catching potential "Structuring" (Transactions between ₹49,000 and ₹49,999)
threshold_min = 49000
threshold_max = 49999

structuring_alerts = df[(df['amount_inr'] >= threshold_min) & (df['amount_inr'] <= threshold_max)]

print(f"🚨 ALERT: Found {len(structuring_alerts)} transactions hovering suspiciously below the ₹50k reporting limit:")
print(structuring_alerts[['transaction_id', 'account_id', 'amount_inr']])
print("-" * 60)

# DETECTOR 2: Catching "Ghost Vendors" (Different vendor names routing money to the EXACT same bank account)
# We group by bank account and count unique vendor names attached to it
vendor_check = df.groupby('destination_bank_acc')['vendor_name'].nunique().reset_index()
suspicious_accounts = vendor_check[vendor_check['vendor_name'] > 1]['destination_bank_acc']

if not suspicious_accounts.empty:
    print("🚨 ALERT: Potential Shell Company / Kickback Scheme Detected!")
    print("The following destination bank accounts are receiving funds from completely different vendor entities:")
    fraud_trail = df[df['destination_bank_acc'].isin(suspicious_accounts)]
    print(fraud_trail[['destination_bank_acc', 'vendor_name', 'amount_inr']])
else:
    print("✅ No duplicate vendor routing anomalies detected.")

# 1. Navigate to your project folder (change this path to where your notebook actually lives)
cd path/to/your/project/folder

# 2. Initialize this folder as a local Git repository
git init

# 3. Add your notebook file to the staging area
git add financial_fraud_investigation.ipynb

# 4. Commit the file with a clear descriptive message
git commit -m "Initial commit: Added financial structuring and time-based fraud detection scripts"

# 5. Rename your default branch to 'main'
git branch -M main

# 6. Link your local repository to your remote GitHub account 
# (Copy the exact line from your open GitHub browser tab!)
git remote add origin https://github.com/your-username/financial-forensic-analytics.git

# 7. Push your code to GitHub
git push -u origin main    
```
